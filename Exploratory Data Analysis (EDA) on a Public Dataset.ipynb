```python
import pandas as pd

df = pd.read_csv("E:/Downloads/archive (2)/tested.csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
missing_values = df.isnull().sum()
print("Missing values per column:")
print(missing_values)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the style for the plots
sns.set_style("whitegrid")

# 1. Histogram of 'Age'
plt.figure(figsize=(10, 6))
sns.histplot(df['Age'].dropna(), kde=True, bins=30)
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()

# 2. Box plot of 'Fare'
plt.figure(figsize=(10, 6))
sns.boxplot(x=df['Fare'])
plt.title('Distribution of Fare')
plt.xlabel('Fare')
plt.show()

# 3. Countplot of 'Survived'
plt.figure(figsize=(8, 5))
sns.countplot(x=df['Survived'])
plt.title('Distribution of Survival (0 = Not Survived, 1 = Survived)')
plt.xlabel('Survived')
plt.ylabel('Count')
plt.show()

# 4. Countplot of 'Pclass'
plt.figure(figsize=(8, 5))
sns.countplot(x=df['Pclass'])
plt.title('Distribution of Passenger Classes')
plt.xlabel('Pclass')
plt.ylabel('Count')
plt.show()

# 5. Countplot of 'Sex'
plt.figure(figsize=(8, 5))
sns.countplot(x=df['Sex'])
plt.title('Distribution of Passenger Gender')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.show()

# 6. Countplot of 'Embarked'
plt.figure(figsize=(8, 5))
sns.countplot(x=df['Embarked'].dropna())
plt.title('Distribution of Embarkation Ports')
plt.xlabel('Embarked')
plt.ylabel('Count')
plt.show()

# 7. Box plots of 'Survived' vs 'Age' and 'Fare'
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
sns.boxplot(x='Survived', y='Age', data=df)
plt.title('Survival by Age')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Age')

plt.subplot(1, 2, 2)
sns.boxplot(x='Survived', y='Fare', data=df)
plt.title('Survival by Fare')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Fare')
plt.tight_layout()
plt.show()

# 8. Countplots of 'Survived' vs 'Pclass', 'Sex', and 'Embarked'
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
sns.countplot(x='Survived', hue='Pclass', data=df)
plt.title('Survival by Pclass')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Count')

plt.subplot(1, 3, 2)
sns.countplot(x='Survived', hue='Sex', data=df)
plt.title('Survival by Sex')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Count')

plt.subplot(1, 3, 3)
sns.countplot(x='Survived', hue='Embarked', data=df)
plt.title('Survival by Embarked')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```
